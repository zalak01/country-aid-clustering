# Project Understanding

## Stakeholders

**Primary stakeholder:** HELP International

**Beneficiary stakeholders:** Countries and vulnerable populations that may receive financial assistance.

## Business Problem

HELP International has raised approximately $10 million and wants to allocate the funds to countries that are in urgent need of assistance.

As a data scientist, my task is to analyse countries using economic, health, and demographic indicators and provide data-driven recommendations. This analysis will help HELP International make better use of the available funds and support its final aid-allocation decision.

## Why Data Analysis and Clustering Are Useful

Data analysis will help us understand the condition of different countries, identify important patterns, study relationships between variables, and recognise countries with extreme economic or health challenges.

Clustering will help us group countries with similar characteristics. After analysing each cluster, we can identify the group of countries that appears to require the most urgent assistance.

## Project Blocks

The project is divided into four blocks:

1. **Data Visualisation**
2. **EDA and Hypothesis Testing**
3. **Machine Learning Modelling**
4. **Deployment**

## Expected Outcome

The expected outcome is to identify and recommend the group of countries that requires urgent assistance, supported by clear analysis and reasons.

The final results will act as a decision-support system for HELP International. The organisation will make the final aid-allocation decision after considering other practical and humanitarian factors.

### Health Expenditure vs Life Expectancy

- A weak positive association is visible between health expenditure as a percentage of GDP and life expectancy.
- The linear trend model has an R-squared of approximately 0.0444.
- The p-value is approximately 0.0063, indicating that the relationship is statistically significant.
- However, health expenditure explains only about 4.4% of the variation in life expectancy, so its practical explanatory power is low.
- Health expenditure as a percentage of GDP alone is not sufficient to explain a country's life expectancy.
- The result shows association, not causation.

### Income vs Life Expectancy

- Income and life expectancy show a statistically significant positive logarithmic association.
- The logarithmic model has an R-squared of approximately 0.6495 and a p-value below 0.0001.
- The model explains about 65% of the observed variation in life expectancy.
- Improvements in life expectancy are greater at lower income levels and gradually reduce as income rises, indicating diminishing returns.
- Equatorial Guinea is a contextual outlier with income of 33,700 and life expectancy of 60.9 years.
- Income alone does not determine life expectancy, and the relationship does not prove causation.

### Life Expectancy Distribution by Income Group

- Median life expectancy increases from the low-income group to the middle- and high-income groups.
- Low-income countries show the greatest variation in life expectancy.
- Haiti is an extreme lower outlier within the low-income group, with life expectancy of approximately 32 years.
- South Africa, Botswana, Namibia, Angola and Congo, Rep. have unusually low life expectancy compared with other countries in the middle-income group.
- Equatorial Guinea and Kazakhstan are lower outliers within the high-income group.
- These differences indicate that income alone does not explain health outcomes and support the use of multiple socioeconomic and health indicators during clustering.
- The income groups and their cut-offs were created only for exploratory analysis and are not official income classifications.

### Income vs Inflation

- Income and inflation show a weak negative association.
- With all 167 countries, the logarithmic trend model has an R-squared of approximately 0.0246 and a p-value of approximately 0.0428.
- Income explains only about 2.5% of the variation in inflation, so it is not a strong predictor.
- Nigeria is an extreme inflation outlier with income of 5,150 and inflation of 104%.
- After excluding Nigeria for sensitivity analysis, the relationship remained weak but statistically significant, showing that Nigeria alone did not create the pattern.
- Timor-Leste, Yemen and Congo, Dem. Rep. fall in the exploratory low-income and above-median-inflation zone, indicating combined economic pressure.
- The chart identifies vulnerable cases but does not prove that low income causes inflation.

### Bottom 10 Countries by GDP per Capita

- Burundi has the lowest GDP per capita in the dataset, at approximately 231.
- All countries in the bottom 10 have GDP per capita below approximately 500, indicating very limited economic output per person.
- The bottom-10 list is concentrated among African countries, showing a regional pattern of low economic capacity in this dataset.
- Low GDP per capita may limit a country's ability to fund healthcare, education, infrastructure and other development programmes independently.
- GDP per capita alone should not determine aid priority; it must be evaluated alongside income, inflation, child mortality, life expectancy and other indicators.
